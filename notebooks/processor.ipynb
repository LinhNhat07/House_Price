```python
import pandas as pd
import numpy as np

# Giả sử df là DataFrame của bạn (đọc từ file hoặc đã có sẵn)
# Ví dụ: df = pd.read_csv('du_lieu_nha.csv')   hoặc   df = pd.read_excel('...')


# 1. Làm sạch cột 'price'
def clean_price(x):
    if pd.isna(x):  # Nếu là NaN thì giữ nguyên để xử lý sau
        return np.nan

    # Chuyển thành chuỗi và loại bỏ khoảng trắng thừa
    x = str(x).strip()

    # Thay thế các ký tự/phrase không mong muốn
    x = (
        x.replace(" tỷVND", "")
        .replace(" tỷ", "")
        .replace(" tyVND", "")
        .replace("tyVND", "")
        .replace("-4%", "")
        .replace("GIẢM GIÁ", "")
        .replace("GIẢM", "")
        .replace("%", "")
        .replace("GIẢM GIÁ GIẢM GIÁ", "")
    )  # đề phòng lặp

    # Loại bỏ dấu phẩy nếu có (thay bằng rỗng hoặc xử lý sau)
    x = x.replace(",", "")

    try:
        # Chuyển thành số thực
        return float(x)
    except (ValueError, TypeError):
        # Nếu không chuyển được → trả về NaN
        return np.nan


# Áp dụng hàm làm sạch
df["price"] = df["price"].apply(clean_price)

# 2. Nếu bạn muốn cột price(m2) cũng sạch tương tự (tuỳ chọn)
# df['price(m2)'] = df['price(m2)'].apply(clean_price)

# 3. Xóa các dòng có giá trị thiếu (NaN) ở các cột quan trọng
# Bạn có thể chọn cột nào cần không được thiếu
important_cols = ["price", "area(m2)", "bedroom", "bathroom", "price(m2)"]

# Xoá dòng nếu bất kỳ cột nào trong important_cols bị NaN
df_clean = df.dropna(subset=important_cols)

# Hoặc nếu muốn linh hoạt hơn, chỉ xoá khi cột price bị NaN
# df_clean = df.dropna(subset=['price'])

# 4. (Tuỳ chọn) Đổi tên cột price thành giá trị đã sạch hơn
# df_clean = df_clean.rename(columns={'price': 'price_tỷ'})

# Xem kết quả
print("Dữ liệu sau khi làm sạch:")
print(
    df_clean[
        ["street", "price", "price(m2)", "area(m2)", "bedroom", "bathroom", "Label"]
    ].head(15)
)

# Kiểm tra xem còn NaN ở cột price không
print(
    "\nSố giá trị NaN trong cột price sau khi làm sạch:", df_clean["price"].isna().sum()
)

# Lưu lại nếu cần
# df_clean.to_csv('nha_sach.csv', index=False, encoding='utf-8-sig')
# df_clean.to_excel('nha_sach.xlsx', index=False)
```

```python
# Xóa nhiều cột cùng lúc
df = df.drop(columns=["description", "code_housing", "date_update"])
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Giả sử df là DataFrame của bạn (đã làm sạch price, xóa cột thừa, v.v.)

# Tạo dictionary để lưu encoder (nếu sau này cần decode lại)
encoders = {}

for col in ["street", "ward", "district","Label"]:
    le = LabelEncoder()
    # Ghi đè trực tiếp vào cột gốc
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le  # lưu encoder để decode nếu cần

# Kiểm tra kết quả
print("Sau khi ghi đè Label Encoding trực tiếp vào cột gốc:")
print(df[["street", "ward", "district", "price (tỷ VND)", "area(m2)"]].head(10))
print("\nKiểu dữ liệu sau encoding:")
print(df[["street", "ward", "district"]].dtypes)

# Nếu muốn decode lại một cột (ví dụ street) về text gốc
# df['street'] = encoders['street'].inverse_transform(df['street'])
```
